**Training accuracy:** mixed effects models and normality of residuals
- see permutation tests for all three due to non-normal residuals in the permutation test section below

```python
# check overall training accuracy 
reg_formula_trainOverall = sm.MixedLM.from_formula(data = both_stacked, formula = 'training ~  condition*group + shape ',groups=both_stacked['sub'])
reg_results_trainOverall = reg_formula_trainOverall.fit()

residuals_train = reg_results_trainOverall.resid
[stats,p] = scipy.stats.normaltest(residuals_train)
print('train overall: ',reg_results_trainOverall.summary(),'normality:',stats,p) #normality of residuals

#check accuracy on the initial training day (day 1 and 2 for the two maps respectively)
reg_formula_trainInitial = sm.MixedLM.from_formula(data = both_stacked, formula = 'firstDayTrain ~  condition*group + shape',groups=both_stacked['sub'])
reg_results_trainInitial = reg_formula_trainInitial.fit()

residuals_train = reg_results_trainInitial.resid
[stats,p] = scipy.stats.normaltest(residuals_train)
print('initial training: ',reg_results_trainInitial.summary(),'normality:',stats,p)

#check accuracy differences for the refresher on day 3
reg_formula_refresher = sm.MixedLM.from_formula(data = both_stacked, formula = 'refAcc ~  condition*group + shape',groups=both_stacked['sub'])
reg_results_refresher = reg_formula_refresher.fit()

residuals_train = reg_results_refresher.resid
[stats,p] = scipy.stats.normaltest(residuals_train)
print('refresher: ',reg_results_refresher.summary(),'normality:',stats,p)
```

**Navigation test accuracy:** mixed effects models and normality of residuals

```python
#overall two-step accuracy
reg_formula = sm.MixedLM.from_formula(data = both_stacked, formula = 'twoStep ~  group*condition + shape',groups=both_stacked['sub'])
reg_results = reg_formula.fit()

residuals = reg_results.resid
[stats,p] = scipy.stats.normaltest(residuals)
print('two-step acc overall:',reg_results.summary(),'normality:',stats,p)

#first step of two-steps
reg_formula = sm.MixedLM.from_formula(data = both_stacked, formula = 'firstStep ~  group*condition + shape', groups=both_stacked['sub'])
reg_results = reg_formula.fit()

residuals = reg_results.resid
[stats,p] = scipy.stats.normaltest(residuals)
print('first step acc:',reg_results.summary(),'normality:',stats,p)

#second step of two-steps
reg_formula = sm.MixedLM.from_formula(data = both_stacked, formula = 'secondStep ~  group*condition + shape', groups=both_stacked['sub'])
reg_results = reg_formula.fit()

residuals = reg_results.resid
[stats,p] = scipy.stats.normaltest(residuals)
print('first step acc:',reg_results.summary(),'normality:',stats,p)


#straight two steps
reg_formula = sm.MixedLM.from_formula(data = both_stacked, formula = 'straightTwoStep ~  group*condition + shape', groups=both_stacked['sub'])
reg_results = reg_formula.fit()

residuals = reg_results.resid
[stats,p] = scipy.stats.normaltest(residuals)
print('straight two step acc:',reg_results.summary(),'normality:',stats,p)

#diagonal two steps
reg_formula = sm.MixedLM.from_formula(data = both_stacked, formula = 'diagTwoStep ~  group*condition + shape', groups=both_stacked['sub'])
reg_results = reg_formula.fit()

residuals = reg_results.resid
[stats,p] = scipy.stats.normaltest(residuals)
print('diag two step acc:',reg_results.summary(),'normality:',stats,p)

#ERRORS 

#same dir error accuracy
reg_formula = sm.MixedLM.from_formula(data = both_stacked, formula = 'same_dir ~ group*condition + shape',groups=both_stacked['sub'])
reg_results = reg_formula.fit()


residuals = reg_results.resid
[stats,p] = scipy.stats.normaltest(residuals)
print('same dir errors:',reg_results.summary(),'normality:',stats,p)



```
